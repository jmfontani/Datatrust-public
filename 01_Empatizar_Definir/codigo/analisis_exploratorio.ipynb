```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Cargar resultados del benchmark
df = pd.read_csv('benchmark_resultados.csv')
df.head()
```

## 1. Métricas generales

```python
# Métricas agregadas
print(f"Total queries evaluadas: {len(df)}")
print(f"\nTiempo promedio: {df['tiempo_min'].mean():.1f} min")
print(f"Tiempo mediano: {df['tiempo_min'].median():.1f} min")
print(f"Tiempo máximo: {df['tiempo_min'].max():.1f} min")

# Tasa de éxito
exitos = (df['encontrado'] == 's').sum()
parciales = (df['encontrado'] == 'p').sum()
print(f"\nTasa de éxito completo: {exitos}/{len(df)} ({exitos/len(df)*100:.0f}%)")
print(f"Tasa parcial: {parciales}/{len(df)} ({parciales/len(df)*100:.0f}%)")
print(f"Tasa de fracaso: {len(df)-exitos-parciales}/{len(df)} ({(len(df)-exitos-parciales)/len(df)*100:.0f}%)")

# Frustración
print(f"\nFrustración promedio: {df['frustracion'].mean():.1f}/5")
print(f"Frustración mediana: {df['frustracion'].median():.1f}/5")
```

## 2. Comparativa por dificultad

```python
# Clasificar por dificultad (basado en IDs)
def clasificar_dificultad(qid):
    if qid.startswith('Q00'):
        return 'Fácil'
    elif qid.startswith('Q01'):
        return 'Medio'
    else:
        return 'Difícil'

df['dificultad'] = df['query_id'].apply(clasificar_dificultad)

# Comparativa
comparativa = df.groupby('dificultad').agg({
    'tiempo_min': ['mean', 'median'],
    'frustracion': 'mean',
    'resultados_relevantes': 'mean'
}).round(2)

print(comparativa)
```

## 3. Visualizaciones

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Tiempo por dificultad
df.boxplot(column='tiempo_min', by='dificultad', ax=axes[0,0])
axes[0,0].set_title('Tiempo de búsqueda por dificultad')
axes[0,0].set_ylabel('Minutos')

# Frustración por dificultad
df.boxplot(column='frustracion', by='dificultad', ax=axes[0,1])
axes[0,1].set_title('Frustración por dificultad')
axes[0,1].set_ylabel('Nivel (1-5)')

# Resultados relevantes
df.boxplot(column='resultados_relevantes', by='dificultad', ax=axes[1,0])
axes[1,0].set_title('Resultados relevantes en top 5')
axes[1,0].set_ylabel('Cantidad')

# Éxito vs Fracaso
exito_counts = df['encontrado'].value_counts()
axes[1,1].pie(exito_counts.values, labels=['Éxito', 'Parcial', 'Fracaso'], autopct='%1.0f%%')
axes[1,1].set_title('Tasa de éxito')

plt.tight_layout()
plt.savefig('benchmark_visualizacion.png', dpi=150)
plt.show()
```
